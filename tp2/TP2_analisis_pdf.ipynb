## 1. Adquisición y metadatos

### 1.1 Descargar el documento

Definir la URL pública y el nombre del archivo local. Descargar con requests o urllib, comprobar que la respuesta sea exitosa y guardar el PDF.

```python
URL_PDF = "https://dialnet.unirioja.es/descarga/articulo/7107375.pdf"
NOMBRE_ARCHIVO = "influenciavideojuegos.pdf"

respuesta = requests.get(URL_PDF)
print("Código de estado HTTP:", respuesta.status_code)

with open(NOMBRE_ARCHIVO, "wb") as f:
    f.write(respuesta.content)

tamano_kb = os.path.getsize(NOMBRE_ARCHIVO) / 1024
print(f"Archivo guardado como '{NOMBRE_ARCHIVO}' ({tamano_kb:.1f} KB)")


pdf_plumber = pdfplumber.open(NOMBRE_ARCHIVO)

print("Tipo de objeto:", type(pdf_plumber))
print("Cantidad de páginas:", len(pdf_plumber.pages))
```

### 1.2 Inspeccionar los metadatos

Abrir el archivo y mostrar título, autor, cantidad de páginas y software de creación. Si un campo no está informado, dejar constancia sin inventarlo.

```python
doc = pymupdf.open(NOMBRE_ARCHIVO)

print("Tipo de objeto:", type(doc))
print("Cantidad de páginas:", doc.page_count)
metadata_mupdf = doc.metadata
metadata_mupdf

doc = pymupdf.open(NOMBRE_ARCHIVO)
metadata_mupdf = doc.metadata

print("Cantidad de páginas:", doc.page_count)

for key in ["title", "author", "creator", "producer"]:
    print(f"{key}: {metadata_mupdf.get(key) or 'No informado'}")
```
